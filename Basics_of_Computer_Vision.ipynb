```python
%matplotlib inline
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("pauch.jpg")
print("Image Shape:",{img.shape})
print("Image Array:")
print(img)
```

```python
img_rgb=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)

plt.figure(figsize=(6,6))
plt.imshow(img_rgb)
plt.axis('off')
plt.title("Loaded Image")
plt.show()
```

```python
img_grayscale = cv2.imread("pauch.jpg",0) # 0 activates gray scale, -1 activates transperancy scale, 1 activates RGB scale
print("Image shape: {img_grayscale.shape}")

#Display
plt.figure(figsize=(6,6))
plt.imshow(img_grayscale, cmap='gray')
plt.axis('off')
plt.title("Grayscale Image")
plt.show()
```

```python
# Red Scale
img = cv2.imread("pauch.jpg")
img[:, :, 1] = 0
img[:, :, 2] = 0

plt.imshow(img)
plt.axis('off')
plt.show()
```

```python
#Green Scale
img = cv2.imread("pauch.jpg")
img[:, :, 0] = 0
img[:, :, 2] = 0

plt.imshow(img)
plt.axis('off')
plt.show()
```

```python
#Blue Scale
img = cv2.imread("pauch.jpg")
img[:, :, 0] = 0
img[:, :, 1] = 0

plt.imshow(img)
plt.axis('off')
plt.show()
```

```python
# Green Blue Scale
img = cv2.imread("pauch.jpg")
img[:, :, 0] = 0

plt.imshow(img)
plt.axis('off')
plt.show()
```

```python
# Red Blue Scale
img = cv2.imread("pauch.jpg")
img[:, :, 1] = 0

plt.imshow(img)
plt.axis('off')
plt.show()
```

```python
# Red Green Scale
img = cv2.imread("pauch.jpg")
img[:, :, 2] = 0

plt.imshow(img)
plt.axis('off')
plt.show()
```
